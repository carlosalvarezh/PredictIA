# **Proyecto: PredictIA**

El **Grupo Argos**, reconocido por su liderazgo en sectores como infraestructura, energía y materiales, planteó una **prueba técnica orientada a evaluar la capacidad de integrar ciencia de datos e inteligencia artificial en procesos de negocio reales**.  

Más allá de un simple ejercicio de modelamiento, la propuesta busca identificar una **visión integral**, capaz de conectar el entendimiento del problema con la implementación de soluciones analíticas reproducibles, escalables y relevantes para la operación corporativa.



## **1. Comprensión del reto**

El desafío técnico se divide en dos componentes complementarios.  

El primero, correspondiente a la **Práctica de Ciencia de Datos y Machine Learning**, consiste en desarrollar un sistema capaz de **pronosticar la demanda** de ciertos insumos y, posteriormente, **clasificar escenarios operativos** en categorías internas denominadas *Alpha* y *Beta*.  

El segundo, relacionado con la **Práctica de Inteligencia Artificial Generativa (GenAI)**, implica diseñar un **agente de asistencia corporativo** que responda consultas sobre políticas internas y datos de gestión humana, determinando si debe consultar una base de conocimientos o información estructurada de un archivo de Excel.  

Ambos componentes tienen como meta común aprovechar la inteligencia artificial como una herramienta confiable para apoyar decisiones estratégicas y operativas, fomentando la trazabilidad, la explicabilidad y la eficiencia tecnológica.


## **2. Enfoque de solución**

Desde una perspectiva analítica, la propuesta se diseñó como un proyecto integral llamado **PredictIA: Sistema de Predicción y Asistencia Inteligente**. El nombre expresa la intención de fusionar el rigor cuantitativo de la analítica clásica con las capacidades adaptativas de la inteligencia artificial moderna. La solución consta de dos módulos principales:  

1. **PredictIA Core (Ciencia de Datos y ML):** dedicado a la predicción de la demanda, a la clasificación *Alpha/Beta* y a ofrecer un servicio de inferencia mediante una API.  

2. **PredictIA GenAI (Agente Conversacional):** diseñado para gestionar información en lenguaje natural, determinando cuándo ofrecer respuestas basadas en conocimientos corporativos y cuándo realizar consultas estructuradas.  

Esta estructura dual facilitará abordar tanto la dimensión **predictiva** del negocio como su componente **cognitivo**, demostrando una visión integral de cómo la inteligencia artificial puede integrarse en distintos niveles de una organización.


## **3. Comprensión del contexto de negocio**

Antes de definir modelos o arquitecturas, el análisis partió de una pregunta fundamental:  

> *¿Qué problemática del Grupo Argos soluciona este sistema?*

En el área de **Abastecimiento**, los analistas son responsables de estimar la demanda y clasificar distintos escenarios operativos para determinar las cantidades a comprar y las estrategias de aprovisionamiento.  

Un error de pronóstico puede traducirse en sobrecostos, inventarios innecesarios o desabastecimiento, mientras que una clasificación incorrecta podría distorsionar las decisiones de planificación.  

Por eso, el valor de PredictIA no solo reside en su precisión estadística, sino también en su **habilidad para proporcionar información confiable, explicable y útil** que apoye decisiones más seguras.  

El modelo debe ser, en esencia, un **asistente técnico de confianza** para los equipos de planeación.


## **4. Diseño arquitectónico y estructura del proyecto**

El diseño técnico responde a tres principios rectores: **orden, trazabilidad y escalabilidad**.  

La estructura del proyecto fue organizada en un repositorio modular con la siguiente lógica:  

* **`data/`**: repositorio local de datos, dividido en crudos, procesados y resultados.  

* **`notebooks/`**: análisis exploratorio, modelado y validaciones.  

* **`src/`**: código productivo dividido en módulos (`features`, `models`, `api`, `scripts`).   

* **`models_artifacts/`**: modelos y transformadores entrenados listos para despliegue.  

* **`reports/`**: reportes, métricas y documentación de resultados.  

* **`genai/`**: implementación del agente conversacional y su lógica de enrutamiento.  

* **`tests/`**: pruebas unitarias y de integración para garantizar la calidad del código.

Esta estructura, más allá de su orden técnico, refleja una **metodología de trabajo reproducible**: cada etapa del proceso analítico, esde la exploración de los datos hasta la comunicación de resultados, está explícitamente documentada y aislada.


## **5. Configuración del entorno y control de versiones**

El proyecto se desarrolló con **Python 3.11**, utilizando un entorno virtual aislado (`.venv`) y una lista de dependencias controlada en `requirements.txt`.  

El archivo `requirements-lock.txt` congela versiones específicas para garantizar la reproducibilidad del entorno.  

El control de versiones se gestiona mediante **Git**, con commits descriptivos y una estructura de ramas (`main` y `feat/*`) que refleja el avance incremental del proyecto.  

El repositorio, alojado localmente durante la etapa inicial, está preparado para publicarse en GitHub una vez se consoliden los resultados, siguiendo buenas prácticas de documentación y trazabilidad.


## **6. Estrategia analítica**

El enfoque metodológico de PredictIA combina precisión técnica y claridad interpretativa.

### **Pronóstico de demanda**

El objetivo no es solo anticipar valores futuros, sino también **cuantificar la incertidumbre** asociada a cada predicción.  

Para ello, se emplearán modelos clásicos de series de tiempo (ETS, SARIMA o Auto-ARIMA), que se compararán mediante **backtesting temporal**.  

Las métricas principales incluyen **MAPE, sMAPE, MAE y MASE**, junto con **intervalos de predicción** que permiten expresar el nivel de confianza del modelo.  

Este componente es esencial para que las áreas de abastecimiento comprendan no solo qué esperar, sino también cuán seguros pueden estar de esas expectativas.


### **Clasificación *Alpha/Beta***

El segundo modelo transformará variables operativas y de entorno en categorías estratégicas (*Alpha*, *Beta*), permitiendo segmentar comportamientos y ajustar políticas.  

Se evaluarán algoritmos basados en árboles (Random Forest, XGBoost) y modelos lineales de referencia.  

La evaluación no se limitará a métricas de exactitud, sino que incluirá **curvas de calibración, matrices de confusión y explicabilidad mediante SHAP**, garantizando la transparencia y la comprensión de los factores que impulsan cada decisión.


### **Servicio de inferencia (API)**

Una vez entrenados, los modelos se integrarán en un servicio **FastAPI**, que permitirá exponerlos de forma controlada y accesible.  


Este servicio aceptará solicitudes en formato JSON, aplicará el pipeline de transformación y clasificación y devolverá resultados en un formato interpretable por otros sistemas.  

La inclusión de un endpoint de salud (`/health`) y de un esquema de validación con **Pydantic** asegura la confiabilidad y el control en la operación.


## **7. Extensión GenAI: un agente corporativo con propósito**

El segundo componente de la prueba aborda el uso de la **Inteligencia Artificial Generativa** en un contexto organizacional.  

La propuesta consiste en un **agente de conversación especializado**, diseñado para decidir si responder una pregunta con información de política interna o consultar un archivo de datos (Excel) con información de cesantías.  

Esta dualidad se resuelve mediante un “decider” que analiza la intención del usuario y selecciona la fuente adecuada, garantizando la **precisión, seguridad y trazabilidad** en cada interacción.  

Aunque este agente se desarrollará de forma simplificada, su diseño contempla la posibilidad de integrarse con servicios en la nube y con modelos de lenguaje más avanzados, lo que permitirá su evolución hacia un entorno productivo controlado y auditable.


## **8. Principios de calidad y valor**

*PredictIA* se sustenta en tres principios esenciales:  

1. **Transparencia:** cada paso del proceso está documentado y las decisiones del modelo son interpretables.  

2. **Reproducibilidad:** la estructura modular y el control de versiones permiten replicar el proyecto en cualquier entorno.  

3. **Relevancia operativa:** los resultados no se presentan como cifras aisladas, sino como información útil para la gestión empresarial.

El proyecto se concibe, por tanto, como una **solución completa**: rigurosa en lo técnico, estructurada en lo metodológico y alineada con las necesidades reales del negocio.

## **10. Extensiones estratégicas y propuesta de valor adicional**

Aunque el alcance del reto está claramente definido, una solución orientada al negocio puede —y debe— ir más allá de la mera ejecución técnica.  

Por ello, el proyecto **PredictIA** incluye una visión prospectiva con posibles líneas de evolución que aportarían valor estratégico al Grupo Argos.


### **a) Análisis causal e impacto de decisiones**

Incorporar técnicas de **inferencia causal** permitiría ir más allá de la correlación entre variables.    

Mientras los modelos de aprendizaje automático responden a la pregunta *“qué ocurrirá”*, los enfoques causales buscan responder *“por qué ocurre”* y *“qué pasaría si cambiáramos una condición del sistema”*.  

En el contexto de abastecimiento, esto permitiría estimar, por ejemplo, el **efecto de políticas de compra, proveedores o estacionalidad** sobre la probabilidad de escenarios *Alpha* o *Beta*.  

El uso de métodos como **Propensity Score Matching**, **IPTW** o **Causal Forests** permitiría medir el impacto de cada variable en condiciones controladas, fortaleciendo la capacidad analítica de la organización para diseñar políticas basadas en evidencia.


### **b) Explicabilidad operativa y confianza del modelo**

El proyecto contempla integrar herramientas de explicabilidad visual (SHAP) y mecanismos de calibración probabilística que faciliten la comunicación de los resultados a los equipos no técnicos.  

El objetivo es que las áreas de negocio comprendan no solo la decisión del modelo, sino también el nivel de confianza y las variables que la determinan.  

De esta manera, PredictIA se convierte en una herramienta de apoyo transparente y auditada, alineada con los principios de ética y responsabilidad en inteligencia artificial.


### **c) Conformal Prediction y umbrales adaptativos**

Como extensión natural de la calibración, se plantea la posibilidad de incorporar técnicas de **Conformal Prediction** que generen intervalos de confianza dinámicos para cada predicción.  

Esto permitiría implementar **umbrales adaptativos**: si la confianza del modelo es baja, la decisión puede escalarse a una revisión humana, garantizando un sistema híbrido de toma de decisiones asistida.


### **d) Escalabilidad y visión cloud**

La arquitectura modular y documentada de PredictIA facilita su futura **orquestación en la nube**, integrando herramientas de versionado de datos (DVC), pipelines automáticos (CI/CD) y despliegues seguros en entornos como Azure, AWS o GCP.  

Esto transformaría el prototipo técnico en una **plataforma analítica corporativa**, alineada con la estrategia digital y la sostenibilidad tecnológica del Grupo Argos.

En conjunto, estas extensiones posicionan a **PredictIA** no solo como una respuesta al reto planteado, sino también como una **base sólida para el desarrollo de inteligencia artificial responsable, explicable y estratégica dentro del ecosistema empresarial**.

## **11. Conclusión**

**PredictIA** representa una aproximación integral a la inteligencia artificial aplicada al negocio.  

No se trata de un ejercicio aislado de modelamiento, sino de una **propuesta metodológica coherente y sostenible**, capaz de demostrar cómo la analítica puede integrarse en procesos empresariales con transparencia, escalabilidad y sentido operativo.  

Más allá de cumplir con los requerimientos técnicos planteados, el proyecto incorpora una visión estratégica orientada a la evolución del sistema.  

Las **extensiones propuestas** —como la inferencia causal para comprender el impacto de las decisiones, la explicabilidad basada en SHAP, la calibración de probabilidades y el uso de la predicción conformal— refuerzan la confianza en los resultados y promueven un enfoque de **inteligencia artificial responsable y auditable**.  

Estas iniciativas complementarias permiten que PredictIA trascienda su papel de prototipo técnico y se proyecte como una **plataforma analítica corporativa**, capaz de integrar modelos predictivos, causalidad, visualización y componentes generativos en una misma estructura.  

De esta forma, el sistema no solo responde a las necesidades actuales del negocio, sino que sienta las bases para una evolución futura hacia entornos híbridos, cloud y con mayor autonomía cognitiva.  

En última instancia, **PredictIA** busca reflejar la madurez técnica, la claridad conceptual y la visión estratégica que requiere una organización de nivel Grupo Argos para **convertir los datos en decisiones inteligentes, explicables y sostenibles en el tiempo**.
